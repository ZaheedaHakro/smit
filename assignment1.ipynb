# Task16

```python
sentence = input("Enter sentence of multiple words : ")
words = set(sentence.split())
frequencies = {}
for word in words:
    freq = 0
    for i in sentence.split():
        if word == i :
            freq += 1
           
    frequencies[word] = freq
    
print(frequencies)      
```

# Task19

```python
students = {}

def add_student(name):
    if name.lower() in students.keys():
        print("Student already present")
        return
    
    students[name.lower()] = None
    print("Student added successfully!")
    

def search_student(name):
    if name.lower() in students.keys():
        print(f"Student Name : {name}")
        print(f"Student Marks : {students[name.lower()]}")
    else:
        print("Student Name does not exists")
    
def enter_marks(name, marks):
    if name.lower() in students.keys():
        students[name.lower()] = marks
    else: 
        print("Student name does not exist") 
        
def calculate_grade(name):
    if name.lower() in students.keys():
        if students[name.lower()] == None:
            return "This Student marks are not given"
        elif students[name.lower()] >= 90:
            return "Grade : A+"
        elif students[name.lower()] >= 80:
            return "Grade : A"
        elif students[name.lower()] >= 70:
            return "Grade : B"
        elif students[name.lower()] >= 60:
            return "Grade : C"
        elif students[name.lower()] >= 50:
            return "Grade : D"
        else:
            return "Grade : FAIL"
        
    else: 
        print("Student name does not exist") 
        
def display():
    Name = "Student Name"
    Marks = "Marks"
    print(f"    {Name:<20}   {Marks:<20}")
    count=1
    for name,marks in students.items():
        if marks == None:
            marks = "-"
        print(f"{count} - {name:<20} : {marks:<20}")   
        count += 1     

print("******MENU******")
print("1 : Add Student")
print("2 : Search Student")
print("3 : Enter/Update Marks")
print("4 : Calculate Grade")
print("5 : Display all Students")
print("Press 0 to exit")
    
while(True):
    
    action = int(input("Enter number of any action you want to perform from menu"))
    
    if action == 1:
        name = input("Enter Student Name ")
        add_student(name)
    elif action == 2:
        name = input("Enter Student Name ")
        search_student(name)
    elif action == 3:
        name = input("Enter Student Name ")
        marks = int(input("Enter Student Marks "))
        
        if marks > 100 or marks < 0:
            print("Please Enter Valid Marks")
        else:
            enter_marks(name,marks)
    elif action == 4 :
        name = input("Enter Student Name ")
        print(calculate_grade(name))
    elif action == 5 :
        display()
    elif action == 0:
        break
    else :
        print("Invalid Action")            

```
